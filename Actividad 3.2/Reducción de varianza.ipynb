```python
#Con varianza corregida, antitetico

def simu(n, N):
    H = []
    for j in range(N):  
        u1 = np.random.random(n)
        x = -np.log(u1)
        y = -np.log(1-u1)
        hi = np.exp(x**2)
        h2 = np.exp(y**2)

        es1 = np.mean(hi)
        es2 = np.mean(h2)
        
        H.append(es1)
        H.append(es2)
        lista = dict(Estimadores = H)
        Estimadores = pd.DataFrame(lista)
    media = np.mean(H)
    var = np.var(H)
    
    return media, var, Estimadores
```

```python
n = 1000   # Por ejemplo
N = 1000  # Por ejemplo
media, varianza, Estimadores = simu(n, N)
print("Media:", media)
print("Varianza:", varianza)
Estimadores
```
